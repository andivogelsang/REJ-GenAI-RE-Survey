# Round 1 vs. Round 2 Comparison

Loads both rounds, reports how their column schemas differ after `genai_re_survey.loading` normalizes them, then renders paired/grouped comparison figures for every RQ1–RQ4 chart plus two general GenAI-usage demographic charts (tool-usage frequency, experience duration).

```python
from genai_re_survey import loading, reports, schema
```

```python
df1 = loading.load_round1()
df2 = loading.load_round2()
df1.shape, df2.shape
```

## Schema diff

What's left after normalization: shared items produce paired bars; anything listed below is round-specific and shown as such (e.g. round 2's new "which GenAI tools do you use" question has no round-1 counterpart).

```python
_ = schema.diff_schemas(df1, df2)
```

## Generate all comparison figures

Saved to `figures/comparison/`.

```python
reports.generate_comparison_report(
    df1, df2,
    outdir="../figures/comparison",
    round1_label="Round 1",
    round2_label="Round 2",
)
```
